```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 1: Load the cleaned dataset

```python
df=pd.read_excel('cleaned_dataset.xlsx')
```

```python
df
```

Step 2: Calculate the total number of Movies and TV
Shows.

```python
df['type'].value_counts()
#6126 movie and 2664 tv shows
```

Step 3: Create visualizations for content distribution.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='type', palette='Set2')

plt.title('Distribution of Movies and TV Shows on Netflix', fontsize=14)
plt.xlabel('Content Type', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Hər sütunun üstünə rəqəmi yaz
for p in plt.gca().patches:
    plt.gca().annotate(f'{int(p.get_height())}',
                       (p.get_x() + p.get_width() / 2, p.get_height()),
                       ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
counts = df['type'].value_counts()

plt.figure(figsize=(7, 7))

colors = ['#66b3ff', '#ff9999']
plt.pie(counts,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        wedgeprops={'edgecolor': 'white', 'linewidth': 2},
        textprops={'fontsize': 12})

plt.title('Content Proportion: Movies vs TV Shows', fontsize=14)
plt.tight_layout()
plt.show()
```

Step 4: Compare content proportions.

```python
x=len(df[df.type=='Movie'])
```

```python
y=len(df[df.type=='TV Show'])
```

```python
print(x/y) #Movies are 2.3 times more than TV Shows
```
