Step 1: Extract and clean country information.

```python
import pandas as pd
df=pd.read_excel('cleaned_dataset.xlsx')
```

```python
df.country.unique()[0:86]
# We see 'Not given', we should delete this rows for true analyze.
```

```python
df_clean = df[df['country'] != 'Not Given']
```

```python
df_clean
```

Step 2: Calculate content count by country.

```python
df_clean.country.value_counts()
```

Step 3: Identify top content-producing countries.

```python
top10=df_clean.country.value_counts().head(10)
```

```python
top10
```

Step 4: Create charts and rankings.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
plt.barh(top10.index, top10.values, color='skyblue')
plt.title('Netflix Content Production by Top 10 Countries', fontsize=16, fontweight='bold')
plt.xlabel('Number of Titles', fontsize=12)
plt.ylabel('Country', fontsize=12)
for index, value in enumerate(top10.values):
    plt.text(value + 5, index, str(value), va='center', fontsize=10)
plt.gca().invert_yaxis()
plt.show()
```

Step 5: Generate business insights.

When we look at the list, it is clear that the U.S. is the leading country by a significant margin; when you sum up the number of films produced by the other 9 countries, it only reaches the number of films produced by the U.S. Seeing Asian countries, especially India and Pakistan, on this list is a result of Netflix’s investment in Asian countries. Considering the large populations of India and Pakistan, it is evident that demand will increase over time, which will also increase the number of films produced, and both India and Pakistan will clearly rise to higher positions on this list.
